### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 🖥️ Vista General (Overview)

Vista General (Overview)

S — Situación:
Andes Retail Group cuenta con operaciones comerciales en Perú, Chile y Colombia durante los años 2024 y 2025. El negocio comercializa productos de Electrónica, Ropa, Deportes y Hogar, por lo que necesita una visión consolidada del desempeño comercial para evaluar ingresos, rentabilidad, volumen de ventas, clientes y pedidos.

C — Complicación:
Aunque la empresa cuenta con datos transaccionales, la información se encontraba dispersa y no permitía identificar rápidamente cómo evolucionaron los ingresos, qué categorías aportan más al negocio, qué países tienen mejor desempeño o qué segmentos de clientes generan mayor rentabilidad.

Q — Pregunta:
¿Cómo ha sido el desempeño comercial general de Andes Retail Group entre 2024 y 2025, y en qué segmentos, categorías o regiones se concentran las principales oportunidades comerciales?

A — Respuesta:
El dashboard muestra que Andes Retail Group generó ingresos totales cercanos a 5,53 millones, con una ganancia total aproximada de 1,94 millones y un margen de ganancia cercano al 35,10%. En la vista general se observa una disminución de ingresos entre 2024 y 2025, lo que indica la necesidad de monitorear la evolución comercial y revisar oportunidades de crecimiento.

#### 🔎 Vista Detalle

S — Situación:
Después de revisar el desempeño general del negocio en la vista Overview, fue necesario profundizar en los resultados para entender qué segmentos de clientes, categorías de producto y regiones explican mejor los ingresos y la ganancia de Andes Retail Group.

C — Complicación:
Aunque el negocio presenta ingresos y rentabilidad positivos, los resultados no se distribuyen de forma uniforme entre segmentos, categorías y regiones. Sin una vista detallada, sería difícil identificar cuáles grupos generan más valor y dónde podrían existir oportunidades de mejora comercial.

Q — Pregunta:
¿Qué segmentos de clientes, categorías de producto y regiones concentran la mayor parte de los ingresos y la ganancia, y dónde pueden existir oportunidades comerciales?

A — Respuesta:
La vista Análisis Detallado muestra que el segmento Premium genera el mayor nivel de ingresos y ganancia, seguido por el segmento Estándar. El segmento Económico tiene una participación menor, por lo que podría requerir una estrategia comercial diferenciada. A nivel de categorías, Deportes y Electrónica presentan un desempeño destacado en ganancia, mientras que Hogar y Ropa muestran resultados ligeramente menores. Por región, Norte lidera en ingresos, aunque Centro y Sur mantienen resultados cercanos. La tabla resumen permite explorar combinaciones específicas por país, región, categoría y segmento para identificar oportunidades comerciales más puntuales.
